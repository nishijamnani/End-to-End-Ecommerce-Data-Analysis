# Order Level EDA

```python
'''
Q1
Question
What is the distribution of order values?
Business Problem
Management wants to understand:
Whether most orders are low-value or high-value
Whether the average order value actually represents a typical customer purchase
This helps in:
Pricing strategy
Discount planning
Understanding customer spending behavior
'''
```

```python
order_items['total_revenue'] = order_items['price'] + order_items['freight_value']
revenue_distribution = order_items.groupby('order_id')[['total_revenue']].sum()
print(revenue_distribution.describe())

bins = [0,400,800,1200,1600]
counts,bins,x = plt.hist(revenue_distribution['total_revenue'],bins,rwidth=0.8)
for i,count in enumerate(counts):
    plt.text(bins[i] + (bins[i+1] - bins[i])/2 , count+600, int(count), ha='center')

plt.show()
```

```python
'''
Q2
Question

Are order values skewed?

Business Problem

If order values are highly skewed:

The average order value may be misleading

Business decisions based only on averages can be incorrect

Understanding skewness helps:

Choose correct KPIs (mean vs median)

Design better dashboards

Avoid wrong conclusions about customer spending
'''
'''
Most customers place low to medium value orders
A small number of orders are extremely expensive
These few high-value orders pull the average upward
This means:
The average order value is misleading if used alone.
'''
revenue_distribution = order_items.groupby('order_id')[['total_revenue']].sum()
print(revenue_distribution.describe())
plt.boxplot(revenue_distribution['total_revenue'])
plt.show()

```

```python
# revenue_distribution[revenue_distribution['total_revenue'] > revenue_distribution['total_revenue'].mean() * 2]
'''
By this logic we are not finding an outlier bcz our data is right skewed means there are most of low value orders and very less high vale orders.
so, any mean based analysis will give the worng output.
'''
```

```python
q1 = revenue_distribution['total_revenue'].quantile(0.25)
q3 = revenue_distribution['total_revenue'].quantile(0.75)
IQR = q3 - q1
upper_bound = q3 + 1.5 * IQR
print(upper_bound)
print(revenue_distribution[revenue_distribution['total_revenue'] > upper_bound])

'''
“Because the majority of orders are low-value, even moderately high orders become statistical outliers. 
This confirms that revenue is highly concentrated and the distribution is heavily right-skewed.”
'''
```

# Product-Level EDA

```python
product_revenue_distribution = order_items.groupby('product_id')[['total_revenue']].sum()
print(product_revenue_distribution.describe())
plt.figure(figsize=(8,4))
plt.boxplot(product_revenue_distribution['total_revenue'])
plt.show()

'''
mean = 480 and median = 168, mean > median = right skewed
min = 9 and max = 67606, high difference means high right skewed.
means that the revenue depends and some product.So, revenue is skewed towards few products.Box plot also shows the high skeweness.
The product-level revenue distribution is highly right-skewed. This is evident because the mean revenue per product is significantly 
higher than the median, and there is a very large gap between the minimum and maximum revenue values.
In other words, overall revenue is concentrated among a few top-performing products rather than being evenly distributed across all products.
'''
```

```python
product_revenue_sorted = product_revenue_distribution.sort_values(
    by='total_revenue',
    ascending=False
)
product_revenue_sorted['cumulative_revenue'] = (
    product_revenue_sorted['total_revenue'].cumsum()
)
total_revenue = product_revenue_sorted['total_revenue'].sum()

product_revenue_sorted['cumulative_revenue_pct'] = (
    product_revenue_sorted['cumulative_revenue'] / total_revenue
) * 100
products_80_percent = product_revenue_sorted[
    product_revenue_sorted['cumulative_revenue_pct'] <= 80
]

print("Number of products contributing to 80% revenue:", products_80_percent.shape[0])
product_revenue_sorted.head(10)
```
